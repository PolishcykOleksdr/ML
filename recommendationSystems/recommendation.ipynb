```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv('products.csv')
df.head()
```

```python
# matrix = df.pivot(index='user_id', columns='product_id', values='rating')
matrix = df.pivot(index='product_id', columns='user_id', values='rating')
matrix = matrix.fillna(0)
print(matrix)
```

```python
product_similarity = cosine_similarity(matrix)

similarity_df = pd.DataFrame(
    product_similarity,
    index=matrix.index,
    columns=matrix.index
)

print(similarity_df)
```

```python
def get_recommended_product(product_id, n_recommendations):
    # product_index = df['product_id'].unique().tolist().index(product_id)
    product_index = matrix.index.get_loc(product_id)
    product_scores = product_similarity[product_index]
    product_idxs = product_scores.argsort()[::-1]

    recommended_products = [df['product_id'].unique()[i] for i in product_idxs if i != product_index]

    return recommended_products[:n_recommendations]
```

```python
product_id = 103
n_of_recommendations = 5
recommendations = get_recommended_product(product_id, n_of_recommendations)

print(f'Top {n_of_recommendations} similar products for product {product_id}')

for i, product in enumerate(recommendations, start=1):
    print(f'Recommendation {i} - product {product}')
```

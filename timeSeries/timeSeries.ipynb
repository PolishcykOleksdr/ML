```python
import pandas as pd
from prophet import Prophet
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('T10Y2Y.csv')
```

```python
df.columns = ['ds', 'y']
df = df.dropna()
```

```python
model = Prophet(daily_seasonality=False, weekly_seasonality=True, yearly_seasonality=True)
model.fit(df)
```

```python
future = model.make_future_dataframe(periods=365)
forecast = model.predict(future)
```

```python
fg1 = model.plot(forecast)
plt.title('Forecast T10Y2Y')
plt.xlabel('Date')
plt.ylabel('Spread (%)')
plt.show()

fg2 = model.plot_components(forecast)
fg2.show()

# First picture - Main prophet prediction
# Second picture - Overall trend
# Third picture - Values between Monday-Friday
# Fourth picture - Seasonal values during year
```

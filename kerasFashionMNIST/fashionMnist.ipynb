```python
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, Flatten, MaxPooling2D, Dense
```

```python
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

x_train = x_train[..., tf.newaxis]
y_train = y_train[..., tf.newaxis]
```

```python
def create_model():
    created_model = Sequential([
        Conv2D(64, (4, 4), activation='relu', input_shape=(28, 28, 1)),
        MaxPooling2D((2, 2)),
        Conv2D(64, (4, 4), activation='relu'),
        MaxPooling2D((2, 2)),
        Flatten(),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])
    return created_model
```

```python
optimizers = ['adam', 'adamw', 'rmsprop']

results = {}

for optimizer in optimizers:
    model = create_model()
    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    model.fit(x_train, y_train, epochs=16, batch_size=256)
    test_loss, test_accuracy = model.evaluate(x_test, y_test)
    results[optimizer] = [test_loss, test_accuracy]
```

```python
for optimizer, (loss, acc) in results.items():
    print('Optimizer -', optimizer)
    print('Loss -', loss)
    print('Accuracy -', acc)
    print('-------------------------------')
```

```python
import spacy
import uk_core_news_sm
# from spacy.lang.uk.examples import sentences
from spacy import displacy
```

```python
nlp = uk_core_news_sm.load()
# doc = nlp(sentences[0])
doc = nlp("Тарас Шевченко народився в Моринцях і працював у Києві.")
print(doc.text)
```

```python
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
displacy.render(doc, style='ent')
```
